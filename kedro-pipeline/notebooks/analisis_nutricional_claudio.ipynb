```
import pandas as pd
import matplotlib.pyplot import plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Configuración de estilo visual
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Carga de Datos (Capa Gold)
Cargamos el archivo Parquet `member2_gold.parquet` que contiene la data procesada.


```
# Asegurarse de correr este notebook desde la raíz del proyecto o apuntar a la ruta correcta
df = pd.read_parquet('../data/03_primary/member2_gold.parquet')
display(df.head())
display(df.info())
```

## 2. Distribución del Índice de Masa Corporal (BMI)
Validaremos cómo se distribuye el BMI en la población y cómo quedó categorizado mediante nuestra lógica de la capa Gold.


```
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

sns.histplot(df['BMXBMI'], kde=True, ax=ax[0], color='skyblue')
ax[0].set_title('Distribución de BMI (Continuo)')
ax[0].set_xlabel('BMI')

sns.countplot(data=df, x='BMI_CATEGORY', order=['Bajo peso', 'Normal', 'Sobrepeso', 'Obesidad'], palette='viridis', ax=ax[1])
ax[1].set_title('Conteo de Pacientes por Categoría de BMI')
ax[1].set_xlabel('Categoría BMI')
ax[1].set_ylabel('Cantidad de Pacientes')

plt.tight_layout()
plt.show()
```

## 3. Matriz de Correlación: Dieta vs Presión Arterial
Analizamos si existe una correlación lineal entre la ingesta calórica/macronutrientes y la presión arterial.


```
cols_of_interest = ['BMXBMI', 'BPXSY1', 'BPXDI1', 'AVG_KCAL', 'AVG_PROT', 'AVG_CARB', 'AVG_FAT']
corr_matrix = df[cols_of_interest].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Matriz de Correlación: Medidas Clínicas vs Nutrición')
plt.show()
```

## 4. Conclusiones y Supuestos Validados
1. **Calidad de Datos:** Los nulos han sido manejados correctamente mediante imputación por mediana en la capa Gold, asegurando un dataset robusto para el backend.
2. **Correlaciones:** (Ejemplo) Observamos una relación positiva entre el BMI y la Presión Sistólica (`BPXSY1`), lo cual justifica las penalizaciones conjuntas que aplicamos en la creación del `cardio_risk_score`.
